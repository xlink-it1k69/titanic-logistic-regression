```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x="Survived")

plt.title("Survival Count")
plt.xlabel("Survived")
plt.ylabel("Number of Passengers")

plt.savefig("../figures/survival_count.png", dpi=300)

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x="Sex")

plt.title("Gender Distribution")

plt.savefig("../figures/gender_distribution.png", dpi=300)

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=df,
    x="Sex",
    hue="Survived"
)

plt.title("Survival by Gender")

plt.savefig("../figures/survival_by_gender.png", dpi=300)

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=df,
    x="Pclass"
)

plt.title("Passenger Class")

plt.savefig("../figures/passenger_class.png", dpi=300)

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=df,
    x="Pclass",
    hue="Survived"
)

plt.title("Survival by Passenger Class")

plt.savefig("../figures/survival_by_class.png", dpi=300)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="Age",
    bins=30,
    kde=True
)

plt.title("Age Distribution")

plt.savefig("../figures/age_distribution.png", dpi=300)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="Fare",
    bins=30,
    kde=True
)

plt.title("Fare Distribution")

plt.savefig("../figures/fare_distribution.png", dpi=300)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="Fare"
)

plt.title("Fare Boxplot")

plt.savefig("../figures/fare_boxplot.png", dpi=300)

plt.show()
```
