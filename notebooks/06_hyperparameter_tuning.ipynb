# 06 - Hyperparameter Tuning

## Objectives

In this notebook we will:

- Optimize Logistic Regression hyperparameters.
- Use GridSearchCV.
- Compare the tuned model with the default model.

Expected outputs

- Best parameters
- Best cross-validation score
- Test accuracy

```python
import os
import sys

sys.path.append(os.path.abspath(".."))
```

```python
import pandas as pd

from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV

from sklearn.metrics import accuracy_score

from src.preprocess import (
    load_data,
    drop_columns,
    fill_missing_values,
    encode_sex,
    encode_embarked
)

from src.train import (
    split_data,
    scale_features
)
```

```python
df = load_data("../data/train.csv")

df = drop_columns(df)

df = fill_missing_values(df)

df = encode_sex(df)

df = encode_embarked(df)

X = df.drop("Survived", axis=1)

y = df["Survived"]

X_train, X_test, y_train, y_test = split_data(X, y)

scaler, X_train_scaled, X_test_scaled = scale_features(
    X_train,
    X_test
)
```

```python
model = LogisticRegression(
    random_state=42,
    max_iter=1000
)
```

```python
param_grid = {
    "C": [0.01, 0.1, 1, 10, 100],
    "solver": [
        "liblinear",
        "lbfgs"
    ]
}
```

```python
grid_search = GridSearchCV(
    estimator=model,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)
```

```python
grid_search.fit(
    X_train_scaled,
    y_train
)
```

```python
print("Best Parameters:")

print(grid_search.best_params_)
```

```python
print(
    f"Best CV Accuracy: {grid_search.best_score_:.4f}"
)
```

```python
best_model = grid_search.best_estimator_

best_model
```

```python
y_pred = best_model.predict(
    X_test_scaled
)

accuracy = accuracy_score(
    y_test,
    y_pred
)

print(f"Test Accuracy: {accuracy:.4f}")
```

```python
comparison = pd.DataFrame({
    "Model": [
        "Default Logistic Regression",
        "Tuned Logistic Regression"
    ],
    "Accuracy": [
        0.8268,      # thay bằng Accuracy thật của bạn ở Bài 9
        accuracy
    ]
})

comparison
```

## Conclusion

- GridSearchCV evaluated multiple combinations of hyperparameters.
- The tuned model achieved the best cross-validation performance.
- The final test accuracy is slightly higher than the default model.
- Hyperparameter tuning can improve model performance, but improvements are not always large.
