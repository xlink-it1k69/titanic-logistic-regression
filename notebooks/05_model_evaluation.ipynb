# 05 - Model Evaluation

## Objectives

In this notebook, we will:

- Evaluate the trained Logistic Regression model.
- Compute classification metrics.
- Visualize the confusion matrix.
- Interpret model performance.

Expected outputs

- Accuracy
- Precision
- Recall
- F1-score
- Confusion Matrix

```python
import os
import sys

sys.path.append(os.path.abspath(".."))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

from src.preprocess import (
    load_data,
    drop_columns,
    fill_missing_values,
    encode_sex,
    encode_embarked
)

from src.train import (
    split_data,
    scale_features,
    train_logistic_regression,
    save_model,
    load_model
)
```

```python
df = load_data("../data/train.csv")

df = drop_columns(df)

df = fill_missing_values(df)

df = encode_sex(df)

df = encode_embarked(df)
```

```python
X = df.drop("Survived", axis=1)

y = df["Survived"]
```

```python
X_train, X_test, y_train, y_test = split_data(X, y)
```

```python
scaler, X_train_scaled, X_test_scaled = scale_features(
    X_train,
    X_test
)
```

```python
model = load_model("../models/logistic_regression.pkl")
```

```python
import joblib

model = joblib.load("../models/logistic_regression.pkl")
```

```python
y_pred = model.predict(X_test_scaled)
```

```python
accuracy = accuracy_score(
    y_test,
    y_pred
)

print(f"Accuracy: {accuracy:.4f}")
```

```python
precision = precision_score(
    y_test,
    y_pred
)

print(f"Precision: {precision:.4f}")
```

```python
recall = recall_score(
    y_test,
    y_pred
)

print(f"Recall: {recall:.4f}")
```

```python
f1 = f1_score(
    y_test,
    y_pred
)

print(f"F1-score: {f1:.4f}")
```

```python
metrics = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1-score"
    ],
    "Score": [
        accuracy,
        precision,
        recall,
        f1
    ]
})

metrics
```

```python
print(
    classification_report(
        y_test,
        y_pred
    )
)
```

```python
cm = confusion_matrix(
    y_test,
    y_pred
)

cm
```

```python
plt.figure(figsize=(6,5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title("Confusion Matrix")

plt.xlabel("Predicted")

plt.ylabel("Actual")

plt.savefig(
    "../figures/confusion_matrix.png",
    dpi=300
)

plt.show()
```

## Observation

- The model correctly classified most passengers.
- There are still some false positives and false negatives.
- Overall performance is good, but there is room for improvement through feature engineering and hyperparameter tuning.
